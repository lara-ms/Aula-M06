## Resumo Teórico

Os testes de hipótese são procedimentos estatísticos inferenciais essenciais para avaliar afirmações e comportamentos sobre uma população a partir de dados coletados em uma amostra representativa. O processo inicia-se com a formulação de duas hipóteses estatísticas mutuamente exclusivas: a **hipótese nula (H₀)**, que representa o estado de referência, a suposição inicial de igualdade ou a ausência de efeito/diferença; e a **hipótese alternativa (H₁)**, que postula uma diferença ou alteração estatisticamente relevante. Utiliza-se a curva normal padronizada e princípios de probabilidade para mensurar a plausibilidade de observar os resultados amostrais caso H₀ fosse verdadeira. A partir do cálculo do Erro Padrão da Média (EP) e do Z-score, determina-se o **p-valor**, confrontando-o com os níveis de significância (α) para fundamentar com rigor a aceitação ou rejeição de H₀.

---


## Exemplo Ilustrativo

Suponha que um estudo de benchmark de infraestrutura em um datacenter de Inteligência Artificial mostra que a vazão média histórica de processamento de requisições de inferência para modelos de linguagem é de 2400 req/s (requisições por segundo). Considere que uma equipe de engenharia implementou uma nova arquitetura de compilação tensorial e paralelismo de pipeline, testando o desempenho em uma amostra de 25 servidores selecionados. Esse grupo apresentou uma taxa média de processamento de 3000 req/s, com um desvio padrão populacional histórico de 1250 req/s. Para testar se o desempenho desse novo grupo de servidores difere significativamente do padrão operacional da infraestrutura, aplica-se o teste da Normal para médias (teste Z), como demonstrado a seguir.

## Dados importantes e suas devidas análises

- Média populacional -> μ = 2400 req/s
- Amostra = 25
- x̄ = 3000 req/s
- Desvio Padrão Populacional = 1250 req/s

– **Média populacional (μ):** é a taxa média verdadeira de todo o parque de servidores históricos sob condições normais de operação (no exemplo, 2400 req/s).

– **Amostra (n):** é o subconjunto de nós de computação selecionados para testar o novo pipeline de aceleração (no exemplo, 25 servidores).

– **Média amostral (x̄):** é a taxa de processamento média calculada diretamente a partir dos dados do grupo de teste (no exemplo, 3000 req/s).

– **Desvio padrão populacional (σ):** mensura o quanto as taxas de processamento individuais variam em torno da média populacional, indicando o grau de dispersão da infraestrutura (no exemplo, 1250 req/s).

---

## Passo 1

H₀ (hipótese nula): a média da amostra é igual à média populacional (x̄ = μ).
→ Ou seja, não há diferença significativa entre o desempenho do cluster otimizado e o padrão populacional.

H₁ (hipótese alternativa): a média da amostra é diferente da média populacional (x̄ ≠ μ).
→ Ou seja, há diferença significativa entre o desempenho do novo cluster e o padrão populacional.

---

## Passo 2

## Cálculo de Z, onde EP é o Erro Padrão da Média.

O EP quantifica a variabilidade esperada da média amostral e avalia a precisão da estimativa calculada.

## Fórmulas Importantes!



Z = (3000 - 2400) / 250 = 2,4





EP = 1250 / √25 -> 250 req/s

---

## Passo 3

## Determinação do parâmetro p.

O valor da área hachurada representa a probabilidade crítica (região de rejeição) que queremos encontrar!


Para isso, executamos os seguintes passos:
- Calculamos o valor da área acumulada de -∞ até 2,4. Chamaremos isso de A1.

A1 = 1 - (valor de pertinência de 2,4) -> 1 - 0,9918 = 0,0082

Importante: Para achar o valor de pertinência (0,9918), basta consultar a tabela de distribuição normal padronizada acumulada P(Z ≤ 2,4).

- Agora calculamos o valor de p (considerando as duas caudas do teste bicaudal):

p = 2 x (1 - 0,9918) = 0.0164 = 1,64%

---

## Passo 4

## Tomada de Decisão.

- Para a tomada de decisão, consultamos na tabela abaixo o nível de significância correspondente ao p-valor e definimos se rejeitamos ou não H0.

No nosso caso, H0 será rejeitado, pois p é igual a 1,64%, ou seja, situa-se no intervalo entre 1% e 5%. Com isso, há elevada consistência e fundamentação estatística para rejeitar H0 em favor de H1.

## Passo 5

## Conclusão.

- Com base no procedimento inferencial realizado, expressamos a conclusão formal no contexto do problema investigado:

No nosso caso, a conclusão ideal seria: "Conclui-se que o throughput médio de processamento do cluster com a nova compilação pode ser considerado estatisticamente diferente (superior) do da infraestrutura geral, a um nível de significância de 1,64%".

## Questão 1

Suponha que testes de telemetria em frotas de veículos autônomos operando em uma cidade inteligente indiquem que o **tempo médio de latência** na inferência de modelos de visão computacional (processamento em tempo real de sensores LiDAR e câmeras para desvio de obstáculos) **é historicamente de 7,0 milissegundos (ms)**. Uma equipe de engenharia desenvolveu uma **nova técnica** de quantização de parâmetros neurais e instalou o modelo otimizado em uma **amostra de 36 veículos autônomos**, registrando sob as mesmas condições de tráfego uma **latência média amostral de 6,0 ms**, com um **desvio padrão populacional histórico de 1,8 ms**.

Para verificar se esse novo modelo de visão computacional apresenta um tempo de resposta estatisticamente diferente da média histórica da frota, aplique o teste da Normal para médias.

- Faça os cálculos detalhados (formulação das hipóteses H₀ e H₁, cálculo do Erro Padrão EP, cálculo do Z-score, determinação do p-valor) e chegue à conclusão técnica do resultado obtido.

---

**Resolução :**

- **PASSO 1** : Formulação das hipóteses

> Hipótese nula (H₀) : μ = 7 ms -> tempo médio do novo modelo ser igual ao historico

> Hipótese alternativa (H₁) : μ ≠ 7 ms -> tempo médio do novo modelo ser diferente do historico

- **PASSO 2** : Cálculo do erro padrão (EP) e Z-score (Z)

> $Z = \frac{x̄ - μ}{σ / \sqrt{n}} = \frac{6 - 7}{1.8 / \sqrt{36}} = \frac{-1}{1.8 / 6} = \frac{-1}{0.3} = - 3.3333$

> $EP = \frac{σ}{\sqrt{n}} = \frac{1,8}{\sqrt{36}} = \frac{1,8}{6} = 0.3 ms$

- **PASSO 3** : Determinação do valor p

> Valor acumulado de ∞ até -3.33 --> P(Z ≥ -3.33) = P(Z ≤ 3.33)

> $A1 = 1 - P(Z ≤ 3.33) = 1 - 0.9995 = 0.0005$

> p = 2 x A1 = 2 x 0.0005 = 0.001 = 0.1%

- **PASSO 4** : Tomada de decisão

> Sendo p-valor = 0.1% e consultando a tabela o nível de significância correspondente ao valor, concluimos que temos uma diferença altamente significante e **certeza máxima de H₀ ser rejeitado**.

- **PASSO 5** : Conclusão

> Conclui-se que o tempo médio de latência no inferência da nova técnica de quantização pode ser considerado estatisticamente diferente (inferior) do da frota geral, a um nível de significância de 0.1%.
